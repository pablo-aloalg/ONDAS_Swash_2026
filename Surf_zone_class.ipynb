**What the columns mean**

| Measurement | Meaning |
|---|---|
| Iribarren | Surf similarity, `slope / sqrt(Hs / L0)`. Reef cases use the fore-reef slope |
| x_break | Where `Hss` has lost 10% of its peak |
| Hss/h at break | Wave height to depth ratio at that point |
| Hss and IG at h = 1 m | Incident and infragravity wave height in the inner surf zone |
| Setup at still-water line | Mean water level where the still water meets the bed |
| Mean shoreline level | Average elevation of the moving shoreline |
| Swash, incident and IG | Vertical size of the shoreline motion in each frequency band |
| Ru2 above still water | Runup exceeded by 2% of the uprushes, measured from the still water level |
| Ru2 + WL | The same runup, measured from the profile datum |

Water levels from the model are measured from the still water level, so a higher `WL` does not appear in `Ru2 above still water`. It does appear in `Ru2 + WL`.
