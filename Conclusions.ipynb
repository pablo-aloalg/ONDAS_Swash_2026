## A. Regular or irregular waves, same energy

```python
show("A: irregular instead of regular")
```

**What we see.** Offshore, breaking and in the inner surf zone the two cases look almost the same: same `Hss`, same setup. Near the shore, the regular case has almost no IG wave, while the irregular sea has an IG wave about as high as the short waves. Runup is several times larger with the irregular sea.

**Why.** Short waves break and their height becomes limited by depth, so the surf zone "forgets" how they arrived. An irregular sea arrives in **groups**. Groups of big waves push the mean level around (bound long wave) and the moving breakpoint generates more long waves. These long waves don't break on a gentle beach; they reach the shoreline and drive the swash.

**For an engineer:** a runup formula or a model with regular waves can underestimate runup on gentle beaches by a factor of several. Use irregular waves.

## B. Bigger waves

```python
show("B: Hs 1 → 4 m")
```

**What we see.** Bigger waves start breaking further offshore. Inside the surf zone, `Hss` at 1 m depth grows much less than `Hs` offshore. Setup grows roughly in proportion to `Hs` (around 0.15 `Hs` at the still-water line). The IG wave grows faster than `Hss`, so `IG/Hss` rises above 1. Runup grows strongly.

**Why.** Broken waves are **depth limited**: in 1 m of water there is only so much wave the depth can hold. The energy the waves lose by breaking does not disappear: the change in radiation stress pushes the mean level up (setup), and the groups feed the long waves, which are not depth limited in the same way.

**For an engineer:** the height of broken waves at the toe of a structure barely tells you about the storm. Setup and IG waves do.

## C. Longer waves

```python
show("C: Tp 7 → 14 s")
```

**What we see.** `Hss` near the shore hardly changes. The IG waves and the runup increase clearly with the period.

**Why.** Longer waves are less steep and shoal more, and longer groups make longer and bigger IG waves. On a gentle slope these long waves are reflected rather than broken, so their energy reaches the shoreline.

**For an engineer:** two storms with the same `Hs` are not the same storm. A long-period swell can flood more than a short windsea of the same height.

## D. Steeper beach

```python
show("D: beach slope 1/30 → 1/15")
```

**What we see.** The surf zone gets narrower: waves break closer to the shore. Setup barely changes. Near the shore the short waves are relatively bigger (`IG/Hss` falls), and runup increases.

**Why.** On a gentle slope (low Iribarren) the short waves are fully broken by the time they reach the shoreline: incident swash is **saturated**. On a steeper slope more of the short-wave motion survives to the shoreline. The Iribarren number, slope / √(Hs/L0), organises this: low means dissipative, IG-dominated swash; high means reflective, incident-dominated swash.

**For an engineer:** runup formulas use the slope and Iribarren for a reason. On steep beaches, the short waves themselves run up.

## E. A sandbar

```python
show("E: sandbar 0 → 2.5 m")
```

**What we see.** With a bar, the waves start breaking on the bar, further offshore. Behind the bar the water gets deeper again, so broken waves can stop breaking and continue smaller towards the beach. `Hss` near the shore is lower than without a bar. The IG waves are reduced much less, so the reduction in runup is smaller than the reduction in short waves.

**Why.** The bar forces breaking in shallow water far from the shore, and the trough behind it lets what is left reform. Long waves are too long to "see" a narrow bar and pass over it.

**For an engineer:** a bar protects against short waves, but much less against flooding by long waves.

## F. Tide on a coral reef

```python
show("F: reef water level −1 → +1 m")
```

**What we see.** Waves break at the reef edge in all three cases. With more water on the reef, more short-wave energy crosses the reef flat: `Hss` on the flat grows with the water level. Runup above still water grows a little. But **`Ru2 + WL`**, the level the water actually reaches, grows much more: by about the 2 m of water level **plus** the extra runup.

**Why.** On the reef flat the waves are depth limited: the depth over the reef decides how big they can be. A higher water level gives a deeper reef, bigger waves on the flat, and a higher water level to start from. The effects add up.

**For an engineer:** on reef coasts, flooding is controlled by the water level (tide, surge, sea-level rise) as much as by the waves. For flooding, the right number is `Ru2 + WL`, not `Ru2`.

## Extra. A rougher reef

```python
show("Extra: reef roughness 0.01 → 0.2")
```

**What we see.** Offshore, in deep water, friction does almost nothing, and breaking at the reef edge doesn't change. On the shallow reef flat, `Hss` decreases with roughness, but the IG waves decrease much more. Setup and runup fall clearly: runup roughly halves.

**Why.** Bed friction matters where the water is shallow and the velocities near the bed are large: on the reef flat. It acts over the whole width of the flat, and it removes energy in proportion to the flow velocity and the time it acts. Long waves move the whole water column for a long time, so friction removes them the most.

**For an engineer:** a healthy, rough reef protects against flooding. Losing reef roughness (coral loss) raises runup even if the reef doesn't get lower.

## The big picture

1. **The surf zone saturates.** After breaking, short-wave height is limited by the depth. Bigger offshore waves don't mean bigger broken waves at the shore.
2. **Breaking moves energy, it doesn't only destroy it.** It raises the mean water level (**setup**, about 0.15–0.2 `Hs`) and feeds **long (IG) waves**.
3. **Wave groups make long waves**, and on gentle beaches long waves dominate the swash. Regular-wave thinking misses most of the runup.
4. **The slope decides what reaches the shoreline:** on gentle beaches IG waves, on steep beaches the short waves too (Iribarren).
5. **For flooding, measure water level plus runup.** On reefs, the water level is the most important control.

### Exit question

> Why can the shoreline flood when the short waves have already broken far offshore?

Because breaking does not remove all the energy at once. It raises the mean water level (setup) and, through the wave groups, makes long infragravity waves that are not depth limited the same way. These reach the shoreline and push the swash much higher than the broken short waves would. Add the tide or surge, and the water reaches places the short waves never could.
